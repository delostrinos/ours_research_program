## Hypothesis

$$\tilde{x}(t) = c_0 + c_1 t + c_2 t^2 + c_3 t^3 + c_4 t^4 + \cdots$$

## True Answer

$$x(t) = t + t^2$$

## True Coefficients

$$c_0 = 0, \quad c_1 = 1, \quad c_2 = 1, \quad c_3 = 0, \quad c_4 = 0, \quad \cdots$$

```python
# Defining the parameters
t_min, t_max = 0, 1
num_of_steps = 100
step_size = t_max / num_of_steps

t_table = []
x_table = []

x0 = 0

t_table.append(t_min)
x_table.append(x0)
```

```python
# Populating list
for i in range (num_of_steps):
  current_t = t_table[-1]
  current_x = x_table[-1]

  slope = 1 + (2 * current_t)
  new_x = current_x + (step_size * slope)
  new_t = current_t + step_size

  t_table.append(new_t)
  x_table.append(new_x)
```

```python
# Vandermonde matrix
import numpy as np
t_array = np.array(t_table)
```

```python
degree = 4
columns = []
```

```python
for n in range(0, degree + 1):
  columns.append(t_array ** n)
```

```python
D = np.column_stack(columns)
```

```python
print(np.round(D, 3))
```

```python
b = np.array(x_table)
b
```

```python
c = np.linalg.lstsq(D,b)
```

```python
# We wanted c1 = 1 and c2 = 1
# Machine essentially rediscovered x(t) = t + t^2
c
```
